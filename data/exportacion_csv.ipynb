```python

import sqlite3
import pandas as pd
from pathlib import Path

# Ajustar a la ruta real de tu base de datos analítica
db_path = Path("cordillera_analytics.sqlite")

# Carpeta de exportación
output_dir = Path("exports_powerbi")
output_dir.mkdir(exist_ok=True)

if not db_path.is_file():
    raise FileNotFoundError(
        f"No se encontró la base de datos: {db_path.resolve()}"
    )

tablas = [
    "fact_detalle_pedidos",
    "dim_cliente",
    "dim_producto",
    "dim_fecha"
]

with sqlite3.connect(db_path) as conn:
    for tabla in tablas:
        df = pd.read_sql_query(
            f'SELECT * FROM "{tabla}"',
            conn
        )

        archivo = output_dir / f"{tabla}.csv"

        df.to_csv(
            archivo,
            index=False,
            encoding="utf-8-sig"
        )

        print(f"{tabla}: {len(df):,} filas exportadas")

print("\nExportación terminada.")
```

```python

df = pd.read_csv(
    output_dir / "fact_detalle_pedidos.csv"
)

entregados = df[df["estado"] == "entregado"]

for columna in [
    "importe_neto_linea_clp",
    "costo_linea_clp",
    "margen_bruto_estimado_linea_clp"
]:
    print(f"{columna}: {entregados[columna].sum():,.0f}")
```
